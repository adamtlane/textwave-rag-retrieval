## Mistral QA Generator Demo Overview

*TextWave* is a modular question-answering system designed to leverage large language models for context-based answer generation. This notebook demonstrates the use of the `QAGeneratorMistral` class, which wraps the MistralAI API to generate answers based *only* on user-specified context. The core logic lives in:

* `textwave/modules/generator/question_answering.py`, which defines the `QAGeneratorMistral` class.

The model is configured via three parameters:

* `api_key` (str): Your **Mistral API key**, loaded from `.env` using `dotenv`.
* `temperature` (float): Controls response creativity.
* `generator_model` (str): Specifies which version of the Mistral model to use (e.g., `"mistral-small-latest"`).

This notebook (`demo_generator.ipynb`) serves as a minimal working example that loads credentials securely from `.env` and defines a small set of context paragraphs. Afterwards, a question is asked that can be answered from that context, and it uses MistralAI to generate a response based on only the context.

It should be noted that the associated `.env` file must define `MISTRAL_API_KEY`. This key can be obtained by signing up at [mistral.ai](https://mistral.ai), navigating to your workspace, and generating a new API key. Be mindful of your Mistral API token limits when running multiple generations.

```python
import sys
import os

project_root = os.path.abspath(os.path.join(os.getcwd(), "..")) # Go up from /notebooks to the project root.

if project_root not in sys.path:
    sys.path.insert(0, project_root)

print("Added to sys.path:", project_root) # Confirm the updated path.
```

```python
from textwave.modules.generator.question_answering import QAGeneratorMistral
from dotenv import load_dotenv

load_dotenv() # Load environment variables from .env file.

api_key = os.getenv("MISTRAL_API_KEY") # Get the API key.

if not api_key:
    raise EnvironmentError("MISTRAL_API_KEY not set. Please create a .env file.")

generator = QAGeneratorMistral(
    api_key=api_key,
    temperature=0.3, # Controls randomness (0 = deterministic, >0 = more creative).
    generator_model="mistral-small-latest" # The model version to use.
) # Instantiate the generator with explicit parameters.

print("QAGeneratorMistral initialized successfully.")
```

```python
context = [
    "Albert Einstein was a theoretical physicist born in Germany.",
    "He developed the theory of relativity, one of the two pillars of modern physics.",
    "Einstein was awarded the Nobel Prize in Physics in 1921 for his explanation of the photoelectric effect."
] # Display context.

question = "What was Einstein awarded the Nobel Prize for?" # Display question.

print("Context:\n\n" + "\n".join(context))
print("\nQuestion:\n\n" + question)
```

```python
answer = generator.generate_answer(query=question, context=context) # Generate the answer using Mistral API.

print("\nGenerated Answer: \n")
print(answer)
```

```python
context_missing = [
    "Einstein loved sailing and often played the violin in his free time.",
    "He had a famously unkempt hairstyle and rarely wore socks."
] # Optional test when the answer is not in the provided context.

question_missing = "What was Einstein awarded the Nobel Prize for?"

answer_missing = generator.generate_answer(query=question_missing, context=context_missing)

print("\nMissing Context Test: \n")
print(answer_missing) # Should respond with "No context".
```
